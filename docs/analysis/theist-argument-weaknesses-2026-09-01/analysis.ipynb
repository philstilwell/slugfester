# Why do the theist sides score lower?

Across 169 matched SLUGFESTER debates and 3,502 scored moves, weak substantiation is a major verified driver: evidence and warrant contributes **1.71 points (27.0%)** of the corrected **6.35-point** mean gap. It is not the whole explanation. Logical coherence contributes **1.81 points**, responsiveness **1.30**, calibration and charity **0.85**, precision **0.52**, and relevance/burden **0.21**.

```python
import importlib.util
import contextlib
import io
from pathlib import Path

script = Path('docs/analysis/theist-argument-weaknesses-2026-09-01/analysis.py')
spec = importlib.util.spec_from_file_location('weakness_analysis', script)
module = importlib.util.module_from_spec(spec)
spec.loader.exec_module(module)
with contextlib.redirect_stdout(io.StringIO()):
    results = module.run_analysis()
dq = results['data_quality']
print(f"Validated {dq['classified_debates']} debates, {dq['scored_moves']} moves, {dq['section_score_checks']} side-sections, and {dq['official_score_checks']} side-overall scores.")
```

```python
primary = results['primary']
print(f"Mean official margin: {primary['mean_margin']:.2f}")
print(f"Outcomes: {primary['non_theist_higher']} non-theist higher / {primary['ties']} ties / {primary['theist_higher']} theist higher")
```

```python
for row in sorted(results['dimension_decomposition'], key=lambda row: row['score_gap_contribution'], reverse=True):
    print(f"{row['label']:<24} contribution={row['score_gap_contribution']:.2f}  share={row['share_of_official_gap']:.1%}")
print(f"{'Rounding residual':<24} contribution={results['rounding_residual']:.2f}")
```

```python
theist = results['evidence_bands']['theist']
non = results['evidence_bands']['non_theist']
print(f"Theist evidence below 70:     {theist['below_70']}/{theist['n_moves']} ({theist['below_70_share']:.1%})")
print(f"Non-theist evidence below 70: {non['below_70']}/{non['n_moves']} ({non['below_70_share']:.1%})")
print(f"Theist evidence below 80:     {theist['below_80']}/{theist['n_moves']} ({theist['below_80_share']:.1%})")
print(f"Non-theist evidence below 80: {non['below_80']}/{non['n_moves']} ({non['below_80_share']:.1%})")
```

```python
checks = {row['subset']: row for row in results['subset_checks']}
for subset, label in [('Constructive moves only', 'Constructives'), ('Replies only', 'Replies'), ('Load-bearing moves only', 'Load-bearing moves'), ('Debates with theist side stored as con', 'Theist stored as con')]:
    row = checks[subset]
    print(f"{label}: evidence gap {row['evidence_warrant_gap']:.2f}; theist/non-theist below-70 shares {row['theist_evidence_below_70_share']:.1%} / {row['non_theist_evidence_below_70_share']:.1%}")
```

## Interpretation

The records support the narrower hypothesis: the theist-side arguments are more often under-substantiated, and the gap survives comparisons within the same move role, among load-bearing moves, and when the theist side is stored as `con`. The data also identify distinct inference and reply-contact failures, so ‘lack of evidence’ alone is incomplete.

They do **not** establish the stronger causal hypothesis that faith commitment produced the weaknesses. The ledgers do not measure private epistemic standards or counterfactual performance.
