# Visitor-interest analysis for SLUGFESTER

This notebook runs the repository-backed analysis used for the accompanying report. Group and panel debates listed in the multi-speaker manifest are excluded from interlocutor appearance counts, but retained in corpus-wide debate statistics.

```python
import json
import subprocess
from pathlib import Path

repo_root = Path.cwd()
script = repo_root / 'docs/analysis/visitor-insights/analyze-visitor-insights.mjs'
completed = subprocess.run(['node', str(script)], cwd=repo_root, check=True, capture_output=True, text=True)
analysis = json.loads(completed.stdout)
analysis
```

## Validation interpretation

All values in `analysis['validation']` should be `True`. The two chart datasets must sum to their stated debate and interlocutor populations.

```python
assert all(analysis['validation'].values())
assert sum(row['debates'] for row in analysis['marginBands']) == analysis['corpus']['debates']
assert sum(row['interlocutors'] for row in analysis['appearanceBands']) == analysis['rankings']['recognizedInterlocutors']
analysis['validation']
```
